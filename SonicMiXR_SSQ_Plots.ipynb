```python
import pandas as pd
import matplotlib.pyplot as plt

# ---------------------------
# 1) Enter the data
# ---------------------------
data = [
    ["P1",  "pre-mix",  1, 0, 0, 1],
    ["P1",  "post-mix", 1, 0, 0, 1],
    ["P2",  "pre-mix",  1, 0, 0, 0],
    ["P2",  "post-mix", 0, 0, 0, 0],
    ["P3",  "pre-mix",  0, 0, 0, 0],
    ["P3",  "post-mix", 0, 0, 0, 0],
    ["P4",  "pre-mix",  1, 1, 1, 1],
    ["P4",  "post-mix", 1, 2, 0, 1],
    ["P5",  "pre-mix",  0, 0, 0, 1],
    ["P5",  "post-mix", 0, 0, 0, 0],
    ["P6",  "pre-mix",  1, 0, 0, 1],
    ["P6",  "post-mix", 1, 0, 0, 1],
    ["P7",  "pre-mix",  0, 0, 0, 1],
    ["P7",  "post-mix", 0, 0, 0, 0],
    ["P8",  "pre-mix",  1, 0, 0, 1],
    ["P8",  "post-mix", 0, 0, 0, 1],
    ["P9",  "pre-mix",  0, 0, 0, 0],
    ["P9",  "post-mix", 0, 0, 0, 0],
    ["P10", "pre-mix",  0, 0, 0, 0],
    ["P10", "post-mix", 0, 0, 0, 0],
]

columns = ["ID", "type", "general_discomfort", "fatigue", "headache", "eye_strain"]
df = pd.DataFrame(data, columns=columns)

# Make sure order is pre first, post second
df["type"] = pd.Categorical(df["type"], categories=["pre-mix", "post-mix"], ordered=True)

# Variables to plot
variables = ["general_discomfort", "fatigue", "headache", "eye_strain"]

# ---------------------------
# 2) Create one plot per variable
# ---------------------------
for var in variables:
    plt.figure(figsize=(6, 4))

    # Plot each participant's pre-post line
    for pid in df["ID"].unique():
        temp = df[df["ID"] == pid].sort_values("type")
        plt.plot(
            temp["type"],
            temp[var],
            marker="o",
            linewidth=1.5,
            alpha=0.6
        )

    # Plot mean line across all participants
    mean_vals = df.groupby("type")[var].mean().reindex(["pre-mix", "post-mix"])
    plt.plot(
        mean_vals.index,
        mean_vals.values,
        marker="o",
        linestyle="--",
        linewidth=3,
        color="black",
        label="Mean"
    )

    plt.title(f"{var.replace('_', ' ').title()} (Pre vs Post)")
    plt.xlabel("Condition")
    plt.ylabel("Score")
    plt.grid(True, alpha=0.3)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

data = [
    ["P1",  "pre-mix",  1, 0, 0, 1],
    ["P1",  "post-mix", 1, 0, 0, 1],
    ["P2",  "pre-mix",  1, 0, 0, 0],
    ["P2",  "post-mix", 0, 0, 0, 0],
    ["P3",  "pre-mix",  0, 0, 0, 0],
    ["P3",  "post-mix", 0, 0, 0, 0],
    ["P4",  "pre-mix",  1, 1, 1, 1],
    ["P4",  "post-mix", 1, 2, 0, 1],
    ["P5",  "pre-mix",  0, 0, 0, 1],
    ["P5",  "post-mix", 0, 0, 0, 0],
    ["P6",  "pre-mix",  1, 0, 0, 1],
    ["P6",  "post-mix", 1, 0, 0, 1],
    ["P7",  "pre-mix",  0, 0, 0, 1],
    ["P7",  "post-mix", 0, 0, 0, 0],
    ["P8",  "pre-mix",  1, 0, 0, 1],
    ["P8",  "post-mix", 0, 0, 0, 1],
    ["P9",  "pre-mix",  0, 0, 0, 0],
    ["P9",  "post-mix", 0, 0, 0, 0],
    ["P10", "pre-mix",  0, 0, 0, 0],
    ["P10", "post-mix", 0, 0, 0, 0],
]

columns = ["ID", "type", "general_discomfort", "fatigue", "headache", "eye_strain"]
df = pd.DataFrame(data, columns=columns)
df["type"] = pd.Categorical(df["type"], categories=["pre-mix", "post-mix"], ordered=True)

variables = ["general_discomfort", "fatigue", "headache", "eye_strain"]

fig, axes = plt.subplots(1, len(variables), figsize=(13, 2.8))
fig.patch.set_facecolor("#d9d9d9")

for ax, var in zip(axes, variables):
    grouped = df.groupby("type", observed=False)[var]
    means = grouped.mean().reindex(["pre-mix", "post-mix"])
    sems = grouped.sem().reindex(["pre-mix", "post-mix"])

    x = [0, 1]
    y = means.values
    e = sems.values

    ax.set_facecolor("#d9d9d9")

    ax.errorbar(
        x, y, yerr=e,
        fmt='none',
        ecolor='black',
        elinewidth=1.2,
        capsize=3,
        capthick=1.2,
        zorder=2
    )
    ax.plot(x, y, color='black', linewidth=1.3, zorder=1)
    ax.scatter(x, y, marker='s', s=28, facecolor='white', edgecolor='black', zorder=3)

    ax.set_title(var.replace("_", " ").title(), fontsize=9, pad=6)
    ax.set_xticks([0, 1])
    ax.set_xticklabels(["Pre", "Post"], fontsize=9)

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)
    ax.spines["left"].set_visible(False)
    ax.spines["bottom"].set_linewidth(1)

    ax.tick_params(axis="y", left=False, labelleft=False)
    ax.tick_params(axis="x", length=0)

    ymin = min(0, np.nanmin(y - e) - 0.05)
    ymax = np.nanmax(y + e) + 0.15
    ax.set_ylim(ymin, ymax)
    ax.set_xlim(-0.2, 1.2)

plt.tight_layout(w_pad=1.5)
plt.show()
```
